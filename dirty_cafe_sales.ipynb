```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('./dirty_cafe_sales.csv')
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
cols=df.columns
cols
```

```python
for col in cols:
    print(f"{col}:")
    print(f"{df[col].unique()}")
    print('-------')
```

```python
summary = []
for col in cols:
    null_counts=df[col].isnull().sum()
    error_counts=(df[col]=='ERROR').sum()
    unknown_counts=(df[col]=='UNKNOWN').sum()
    total_counts= null_counts +  error_counts + unknown_counts
    summary.append([col, null_counts, error_counts, unknown_counts, total_counts])
summary_df=pd.DataFrame(summary, columns=['Column', 'Blank', 'ERROR', 'UNKNOWN', 'Total Counts'])
summary_df
```

```python
num_cols=['Quantity','Price Per Unit', 'Total Spent']
for col in num_cols:
    df[col]=df[col].replace(['ERROR', 'UNKNOWN'], np.nan)
    df[col] = pd.to_numeric(df[col], errors='coerce')

print(df[num_cols].isnull().sum())
```

```python
for col in num_cols:
    print(f"{col}:")
    print(f"{df[col].unique()}")
    print('-------')
```
